```python
# Task 01 – Adaptive Search Agent for Dynamic Environments

import random
import time
from collections import deque

print("==== Adaptive Search Agent (Dynamic Grid) ====\n")

# Grid settings
GRID_SIZE = 6
WALL_PROBABILITY = 0.2
SCENARIOS = 50

START = (0, 0)
GOAL = (5, 5)

# -----------------------------
# Generate random grid
# -----------------------------
def generate_grid():
    grid = []
    for i in range(GRID_SIZE):
        row = []
        for j in range(GRID_SIZE):
            row.append(1 if random.random() < WALL_PROBABILITY else 0)
        grid.append(row)

    grid[START[0]][START[1]] = 0
    grid[GOAL[0]][GOAL[1]] = 0
    return grid

# -----------------------------
# Breadth First Search
# -----------------------------
def bfs(grid, start, goal):
    queue = deque([(start, [start])])
    visited = set([start])

    while queue:
        (x, y), path = queue.popleft()

        if (x, y) == goal:
            return path

        for dx, dy in [(-1,0), (1,0), (0,-1), (0,1)]:
            nx, ny = x + dx, y + dy
            if 0 <= nx < GRID_SIZE and 0 <= ny < GRID_SIZE:
                if grid[nx][ny] == 0 and (nx, ny) not in visited:
                    visited.add((nx, ny))
                    queue.append(((nx, ny), path + [(nx, ny)]))

    return None

# -----------------------------
# Change environment dynamically
# -----------------------------
def change_environment(grid):
    x, y = random.randint(0, 5), random.randint(0, 5)
    if (x, y) != START and (x, y) != GOAL:
        grid[x][y] = 1 - grid[x][y]

# -----------------------------
# Simulation and evaluation
# -----------------------------
success = 0
total_path_length = 0
total_replan_time = 0

for scenario in range(1, SCENARIOS + 1):
    print(f"\n--- Scenario {scenario} ---")
    print("Start:", START, "Goal:", GOAL)

    grid = generate_grid()
    agent_pos = START

    path = bfs(grid, agent_pos, GOAL)

    if not path:
        print("Result: No initial path found.")
        continue

    for step in path[1:]:
        # Random environment change
        if random.random() < 0.3:
            change_environment(grid)

            replan_start = time.time()
            new_path = bfs(grid, agent_pos, GOAL)
            total_replan_time += time.time() - replan_start

            if not new_path:
                print("Result: Replanning failed.")
                break

            path = new_path

        agent_pos = step

        if agent_pos == GOAL:
            success += 1
            total_path_length += len(path)
            print("Result: Goal reached.")
            break

# -----------------------------
# Performance summary
# -----------------------------
print("\n==== Performance Evaluation ====")
print("Total Scenarios:", SCENARIOS)
print("Successful Runs:", success)
print("Success Rate:", round((success / SCENARIOS) * 100, 2), "%")

if success > 0:
    print("Average Path Length:", round(total_path_length / success, 2))
    print("Average Replanning Time:", round(total_replan_time / success, 4), "seconds")
```

```python
# Task 02: Mini Expert System for Cybersecurity Threat Diagnosis

print("==== Mini Expert System: Cybersecurity Threat Diagnosis ====\n")

# Define rules as a list of dictionaries (if conditions then conclusion)
rules = [
    {
        "conditions": {"high_traffic": True, "no_user_activity": True},
        "conclusion": "DDoS Attack"
    },
    {
        "conditions": {"multiple_ports_accessed": True, "unusual_scan_pattern": True},
        "conclusion": "Port Scan Attack"
    },
    {
        "conditions": {"malicious_file_detected": True},
        "conclusion": "Malware Infection"
    },
    {
        "conditions": {"login_failures": True, "multiple_ips": True},
        "conclusion": "Brute Force Attack"
    },
    {
        "conditions": {"data_leak_detected": True},
        "conclusion": "Data Breach"
    }
]

# Function to ask user for a boolean input (yes/no)
def ask_user(question):
    while True:
        answer = input(question + " (yes/no): ").strip().lower()
        if answer in ['yes', 'y']:
            return True
        elif answer in ['no', 'n']:
            return False
        else:
            print("Please answer 'yes' or 'no'.")

# Forward chaining inference engine
def diagnose(threat_facts):
    reasoning_trace = []
    for rule in rules:
        conditions = rule["conditions"]
        # Check if all conditions for this rule are met
        match = True
        for cond, val in conditions.items():
            if cond not in threat_facts:
                # Ask user for missing fact
                threat_facts[cond] = ask_user(f"Is it true that {cond.replace('_', ' ')}?")
                reasoning_trace.append(f"Asked user about '{cond}': {threat_facts[cond]}")
            if threat_facts[cond] != val:
                match = False
                reasoning_trace.append(f"Condition '{cond}' = {threat_facts[cond]} (expected {val}) -> no match")
                break
            else:
                reasoning_trace.append(f"Condition '{cond}' = {threat_facts[cond]} (expected {val}) -> matched")

        if match:
            reasoning_trace.append(f"Rule matched: Diagnosed as {rule['conclusion']}")
            return rule["conclusion"], reasoning_trace

    reasoning_trace.append("No matching threat diagnosis found.")
    return "Unknown Threat", reasoning_trace

# Test scenarios (simulate known facts)
test_scenarios = [
    {"high_traffic": True, "no_user_activity": True},                  # DDoS
    {"multiple_ports_accessed": True, "unusual_scan_pattern": True},  # Port Scan
    {"malicious_file_detected": True},                                # Malware
    {"login_failures": True, "multiple_ips": True},                   # Brute Force
    {"data_leak_detected": True},                                     # Data Breach
    {"high_traffic": False, "no_user_activity": True},                # No known threat
    {"multiple_ports_accessed": False, "unusual_scan_pattern": True}, # No known threat
    {"login_failures": True, "multiple_ips": False},                  # No known threat
    {"malicious_file_detected": False},                               # No known threat
    {}                                                                # All unknown, user will answer
]

correct_results = [
    "DDoS Attack",
    "Port Scan Attack",
    "Malware Infection",
    "Brute Force Attack",
    "Data Breach",
    "Unknown Threat",
    "Unknown Threat",
    "Unknown Threat",
    "Unknown Threat",
    None  # last one depends on user answers
]

print("Running test scenarios...\n")

correct = 0
for i, scenario in enumerate(test_scenarios):
    print(f"--- Scenario {i+1} ---")
    diagnosis, trace = diagnose(scenario.copy())  # use copy to not mutate original

    print("Diagnosis:", diagnosis)
    print("Reasoning Trace:")
    for step in trace:
        print("  -", step)

    # Check accuracy except last scenario (user dependent)
    if correct_results[i] is not None:
        if diagnosis == correct_results[i]:
            correct += 1
            print("Result: Correct\n")
        else:
            print(f"Result: Incorrect (Expected: {correct_results[i]})\n")
    else:
        print("Result: Depends on user input, no accuracy check.\n")

accuracy = (correct / (len(test_scenarios) - 1)) * 100
print(f"Accuracy on 9 automatic scenarios: {accuracy:.2f}%")

print("\n==== End of Expert System ====")
```
