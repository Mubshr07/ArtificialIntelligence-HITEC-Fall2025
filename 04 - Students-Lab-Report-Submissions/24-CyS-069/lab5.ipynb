```python
import numpy as np

arr = np.array([[0, 1], [2, 3]])
print("Original array:\n", arr)

flat_arr = arr.flatten()

print("Maximum value of the above flattened array:", flat_arr.max())
print("Minimum value of the above flattened array:", flat_arr.min())
```

```python
import numpy as np

matrix = np.arange(2, 11).reshape(3, 3)
print("3x3 matrix with values from 2 to 10:\n", matrix)
```

```python
import numpy as np

arr = np.array([1, 2, 3, 4])
float_arr = arr.astype(float)

print("Original array:", arr)
print("Array converted to a float type:", float_arr)
```

```python
import numpy as np

M1 = np.array([[5, 5], [0, 1]])
M2 = np.array([[1, 1], [2, 1]])

# Matrix multiplication
result = np.dot(M1, M2)

# Transpose of the result
transpose_result = result.T

print("Matrix M1:\n", M1)
print("Matrix M2:\n", M2)
print("Result of multiplication:\n", result)
print("Transpose of resultant matrix:\n", transpose_result)
```

```python
import numpy as np

A = np.array([
    [6, 1, 1, 3],
    [4, -2, 5, 1],
    [2, 8, 7, 6],
    [3, 1, 9, 7]
])

# Compute inverse
A_inv = np.linalg.inv(A)

print("Original Matrix A:\n", A)
print("Inverse of Matrix A:\n", A_inv)
```
