```python
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius ** 2

    def perimeter(self):
        return 2 * 3.14 * self.radius

r = float(input("Enter radius: "))
c = Circle(r)

print("Area:", c.area())
print("Perimeter:", c.perimeter())
```

```python
class Power:
    def pow(self, x, n):
        return x ** n

x = int(input("Enter base: "))
n = int(input("Enter exponent: "))

p = Power()
print("Result:", p.pow(x, n))
```

```python
class Animal:
    def speak(self):
        print("Animal Speaking")

class Dog(Animal):
    def bark(self):
        print("Dog Barking")

class Puppy(Dog):
    def eat(self):
        print("Puppy Eating")

p = Puppy()
p.speak()
p.bark()
p.eat()
```
