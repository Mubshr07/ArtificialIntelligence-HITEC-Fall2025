```python
import random
import time


class Environment:
    def __init__(self):
        self.rooms = {
            'A': random.choice(['Clean', 'Dirty']),
            'B': random.choice(['Clean', 'Dirty']),
            'C': random.choice(['Clean', 'Dirty'])
        }


class SimpleReflexAgent:
    def __init__(self, environment):
        self.env = environment
        self.location = random.choice(list(self.env.rooms.keys()))
        self.clean_counter = 0

    def perceive(self):
        return self.location, self.env.rooms[self.location]

    def clean(self):
        print(f"Cleaning room {self.location}...")
        self.env.rooms[self.location] = 'Clean'
        self.clean_counter += 1

    def move(self):
        if all(state == 'Clean' for state in self.env.rooms.values()):
            print("All rooms are clean. Moving randomly...")
        self.location = random.choice(list(self.env.rooms.keys()))

    def run(self, steps=10):
        for _ in range(steps):
            location, status = self.perceive()
            print(f"Location: {location}, Status: {status}")

            if status == 'Dirty':
                self.clean()
            else:
                self.move()

            print(f"Current room states: {self.env.rooms}")
            print(f"Clean counter: {self.clean_counter}\n")
            time.sleep(1)


env = Environment()
agent = SimpleReflexAgent(env)
agent.run(10)
```

```python
class SmartLightAgent: 
    def __init__(self): 
        self.state = "OFF" 
 
    def perceive_and_act(self, motion, light_level, time_of_day): 
        if motion and (light_level == "Dark" or time_of_day == "Night"): 
            self.state = "ON" 
        elif not motion or light_level == "Bright": 
            self.state = "OFF" 
        return self.state 
 
agent = SmartLightAgent() 
 
scenarios = [ 
    {"motion": True, "light_level": "Dark", "time_of_day": "Night"}, 
    {"motion": False, "light_level": "Bright", "time_of_day": "Day"}, 
    {"motion": True, "light_level": "Bright", "time_of_day": "Day"}, 
    {"motion": False, "light_level": "Dark", "time_of_day": "Night"}, 
] 
 
for s in scenarios: 
    state = agent.perceive_and_act(s["motion"], s["light_level"], 
s["time_of_day"]) 
    print(f"Percepts: {s} → Light: {state}") 
```

```python
import random


class ThermostatAgent:
    def __init__(self):
        self.past_temps = []
        self.current_temp = random.randint(18, 26)
        self.state = "Idle"

    def sense(self):
        self.current_temp += random.choice([-1, 0, 1])
        self.past_temps.append(self.current_temp)
        if len(self.past_temps) > 5:
            self.past_temps.pop(0)

    def analyze_trend(self):
        if len(self.past_temps) < 2:
            return "Stable"

        diff = self.past_temps[-1] - self.past_temps[0]

        if diff > 2:
            return "Rising Fast"
        elif diff < -2:
            return "Falling Fast"
        else:
            return "Stable"

    def act(self):
        trend = self.analyze_trend()

        if self.current_temp < 20 or trend == "Falling Fast":
            self.state = "Heating"
        elif self.current_temp > 24 or trend == "Rising Fast":
            self.state = "Cooling"
        else:
            self.state = "Idle"

    def run(self, steps=10):
        for _ in range(steps):
            self.sense()
            self.act()
            print(
                f"Temp: {self.current_temp}°C | "
                f"Trend: {self.analyze_trend()} | "
                f"Action: {self.state}"
            )


agent = ThermostatAgent()
agent.run(15)
```
