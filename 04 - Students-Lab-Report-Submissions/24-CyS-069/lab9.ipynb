```python
import os
import pandas as pd
import matplotlib.pyplot as plt

os.chdir(r'D:\destop\artificial intelligence\lab artificial intelligence')

df = pd.read_csv('heart_disease.csv.txt')
df.head()
```

```python
# Mean age
mean_age = df['Age'].mean()
print("Mean Age:", mean_age)

```

```python
# Heart disease count
heart_disease_count = df['Heart Disease Status'].value_counts()
print("\nHeart Disease Count:\n", heart_disease_count)

```

```python
# Cholesterol
max_chol = df['Cholesterol Level'].max()
min_chol = df['Cholesterol Level'].min()
print("\nMax Cholesterol:", max_chol)
print("Min Cholesterol:", min_chol)

```

```python
# BMI over 30
bmi_over_30 = df[df['BMI'] > 30].shape[0]
print("\nIndividuals with BMI > 30:", bmi_over_30)

```

```python
# High BP and heart disease
high_bp_hd = df[
    (df['Blood Pressure'] > 140) &
    (df['Heart Disease Status'] == 'Yes')
].shape[0]
print("\nHigh BP & Heart Disease:", high_bp_hd)

```

```python
# Histogram BMI
plt.figure()
plt.hist(df['BMI'], bins=10, edgecolor='black')
plt.title('BMI Distribution')
plt.xlabel('BMI')
plt.ylabel('Frequency')
plt.show()

```

```python
# Bar chart exercise vs heart disease
exercise_hd = df.groupby('Exercise Habits')['Heart Disease Status'].value_counts().unstack()
exercise_hd.plot(kind='bar')
plt.title('Heart Disease vs Exercise Habits')
plt.ylabel('Number of Individuals')
plt.show()

```

```python
# Lowest risk group
lowest_risk = df[df['Heart Disease Status'] == 'No']['Exercise Habits'].value_counts()
print("\nLowest Risk Group:\n", lowest_risk)

```

```python
# Pie chart smokers
plt.figure()
df['Smoking Status'].value_counts().plot(
    kind='pie', autopct='%1.1f%%', startangle=90
)
plt.title('Smoking Distribution')
plt.ylabel('')
plt.show()

```

```python
# Smoking percentage
smoker_percentage = df['Smoking Status'].value_counts(normalize=True) * 100
print("\nSmoking Percentage:\n", smoker_percentage)

```

```python
# Scatter plot BMI vs Triglycerides
plt.figure()
plt.scatter(df['BMI'], df['Triglyceride Level'], alpha=0.6)
plt.title('BMI vs Triglyceride Level')
plt.xlabel('BMI')
plt.ylabel('Triglyceride Level')
plt.show()

```

```python
# Box plot sleep hours
plt.figure()
df.boxplot(column='Sleep Hours')
plt.title('Sleep Hours Distribution')
plt.show()

```

```python
# Median sleep hours
median_sleep = df['Sleep Hours'].median()
print("\nMedian Sleep Hours:", median_sleep)


```
