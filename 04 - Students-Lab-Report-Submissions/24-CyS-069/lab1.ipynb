```python
 
num = int(input("Enter a number: "))

n = num
digit_sum = 0

while n > 0:
    digit_sum += n % 10
    n //= 10

if digit_sum != 0 and num % digit_sum == 0:
    print(f"{num} is a Harshad Number.")
else:
    print(f"{num} is not a Harshad Number.")

 
 
```
