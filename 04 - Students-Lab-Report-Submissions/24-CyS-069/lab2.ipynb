```python
sample_list = ['abc', 'xyz', 'aba', '1221', 'xyzzyx', 'aa', '122']

count = 0
for s in sample_list:
    if len(s) >= 2 and s[0] == s[-1]:
        count += 1

print("Expected Result:", count)
```

```python
nums = (1, 2, 4, 2, 6)

sum_nums = sum(nums)

product = 1
for n in nums:
    product *= n

print("Sum is:", sum_nums)
print("Product is:", product)
```

```python
num_list = [1, 2, 4, 2, 6]

largest = max(num_list)

print("Largest number is:", largest)
```

```python
L = [(), (), ('',), ('a', 'b'), (), ('a', 'b', 'c'), (), ('d')]

result = [t for t in L if t]

print("Result:", result)
```

```python
n = int(input("Enter a number for dictionary generation: "))

d = {str(i): i * i for i in range(1, n + 1)}

print("Generated dictionary:", d)
```
