```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)

num = int(input("Enter a number: "))
print("Factorial:", factorial(num))
```

```python
def unique_list(lst):
    return list(set(lst))

sample_list = [1, 2, 3, 3, 4, 4, 5, 3, 7, 5]
print("Unique List:", unique_list(sample_list))
```

```python
def even_numbers(n):
    evens = [2 * i for i in range(1, n + 1)]
    return evens

n = int(input("Enter number of even terms: "))
evens = even_numbers(n)

sum_evens = sum(evens)

prod = 1
for e in evens:
    prod *= e

print("Even numbers:", evens)
print("Sum:", sum_evens)
print("Product:", prod)
```

```python
def match_strings(str1, str2):
    if str1 == str2:
        return "Strings match!"
    else:
        return "Strings do not match!"

a = input("Enter first string: ")
b = input("Enter second string: ")
print(match_strings(a, b))
```
