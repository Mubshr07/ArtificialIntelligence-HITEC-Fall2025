```python
import heapq 
 
graph_a = { 
    'a': {'b': 9, 'c': 4, 'd': 7}, 
    'b': {'a': 9, 'c': 14, 'e': 11}, 
    'c': {'a': 4, 'b': 14, 'd': 18, 'f': 12}, 
    'd': {'a': 7, 'c': 18, 'f': 14}, 
    'e': {'b': 11, 'z': 5}, 
    'f': {'c': 12, 'd': 14, 'z': 9}, 
    'z': {} 
} 
 
heuristic_a = { 
    'a': 21, 'b': 14, 'c': 17, 'd': 18, 
    'e': 5, 'f': 8, 'z': 0 
} 
 
def best_first_search(graph, h, start, goal): 
    visited = set() 
    pq = [(h[start], start)]   
    path = [] 
 
    while pq: 
        _, node = heapq.heappop(pq) 
        path.append(node) 
        if node == goal: 
            return path 
        visited.add(node) 
 
        for neighbor in graph[node]: 
            if neighbor not in visited: 
                heapq.heappush(pq, (h[neighbor], neighbor)) 
 
    return None 
 
def beam_search(graph, h, start, goal, beam_width=2): 
    frontier = [start] 
    path = [start] 
 
    while frontier: 
        candidates = [] 
 
        for node in frontier: 
            if node == goal: 
                return path + [goal] 
 
            for neighbor in graph[node]: 
                candidates.append((h[neighbor], neighbor)) 
 
        candidates.sort() 
        frontier = [n for (_, n) in candidates[:beam_width]] 
        path += frontier 
 
    return None 
 
 
print("Task (a) — Best First Search:", best_first_search(graph_a, heuristic_a, 'a', 'z')) 
print("Task (a) — Beam Search:", beam_search(graph_a, heuristic_a, 'a', 'z')) 
```

```python
import heapq 
 
graph_b = { 
    'S': {'A': 4, 'B': 8, 'C': 11}, 
    'A': {'D': 5}, 
    'B': {'D': 15, 'F': 6}, 
    'C': {'E': 20, 'F': 8}, 
    'D': {'H': 16, 'I': 1}, 
    'E': {'G': 19}, 
    'F': {'G': 13, 'I': 5}, 
    'H': {'J': 2}, 
    'I': {'J': 3, 'K': 15}, 
    'J': {}, 
    'K': {'G': 16}, 
    'G': {} 
} 
 
heuristic_b = { 
    'S':7, 'A':8, 'B':6, 'C':5, 'D':6, 'E':5, 
    'F':3, 'G':0, 'H':7, 'I':4, 'J':5, 'K':3 
} 
 
def best_first(graph, h, start, goal): 
    visited = set() 
    pq = [(h[start], start)] 
    path = [] 
 
    while pq: 
        _, node = heapq.heappop(pq) 
        path.append(node) 
        if node == goal: 
            return path 
        visited.add(node) 
 
        for neighbor in graph[node]: 
            if neighbor not in visited: 
                heapq.heappush(pq, (h[neighbor], neighbor)) 
 
    return None 
 
def beam(graph, h, start, goal, width=2): 
    frontier = [start] 
    path = [start] 
 
    while frontier: 
        candidates = [] 
 
        for node in frontier: 
            if node == goal: 
                return path 
 
            for neighbor in graph[node]: 
                candidates.append((h[neighbor], neighbor)) 
 
        candidates.sort() 
        frontier = [node for (_, node) in candidates[:width]] 
        path += frontier 
 
    return None 
 
def a_star(graph, h, start, goal): 
    pq = [(0 + h[start], 0, start, [])]  # (f, g, node, path) 
    visited = set() 
 
    while pq: 
        f, g, node, path = heapq.heappop(pq) 
 
        if node == goal: 
            return path + [node] 
 
        if node in visited: 
            continue 
 
        visited.add(node) 
 
        for neighbor in graph[node]: 
            cost = g + graph[node][neighbor] 
            heapq.heappush(pq, (cost + h[neighbor], cost, neighbor, path + [node])) 
 
    return None 
 
print("\nTask (b) — Best First Search:", best_first(graph_b, heuristic_b, 'S', 'G')) 
print("Task (b) — Beam Search:", beam(graph_b, heuristic_b, 'S', 'G')) 
print("Task (b) — A* Search:", a_star(graph_b, heuristic_b, 'S', 'G')) 
```
