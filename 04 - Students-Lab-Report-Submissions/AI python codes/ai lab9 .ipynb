```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("heart_disease.csv")

print("Dataset Preview:")
print(df.head())

# 1) Heart Disease count
print("\nHeart Disease Status Count:")
print(df['Heart Disease Status'].value_counts())

# 2) Maximum & Minimum Cholesterol
print("\nCholesterol Level:")
print("Max:", df['Cholesterol Level'].max())
print("Min:", df['Cholesterol Level'].min())

# 3) Individuals with BMI over 30
print("\nIndividuals with BMI > 30:")
print(df[df['BMI'] > 30].shape[0])

# 4) High Blood Pressure AND Heart Disease
print("\nHigh Blood Pressure and Heart Disease:")
print(df[(df['High Blood Pressure'] == 1) & (df['Heart Disease Status'] == 1)].shape[0])

# 5) Histogram of BMI
plt.hist(df['BMI'], bins=10)
plt.xlabel("BMI")
plt.ylabel("Frequency")
plt.title("BMI Distribution")
plt.show()

# 6) Exercise Habits vs Heart Disease (Bar Chart)
df[df['Heart Disease Status'] == 'Yes'].groupby('Exercise Habits').size().plot(kind='bar')
plt.xlabel('Exercise Habits')
plt.ylabel('Number of Heart Disease Cases')
plt.title('Heart Disease Cases by Exercise Habits')
plt.show()


# 7) Lowest risk group by exercise
print("\nHeart Disease Risk by Exercise Habits:")
print(df[df['Heart Disease Status'] == 'Yes'].groupby('Exercise Habits').size())


# 8) Smokers vs Non-Smokers (Pie Chart)
df['Smoking'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title("Smokers vs Non-Smokers")
plt.ylabel("")
plt.show()

# 9) Percentage of smokers
print("\nPercentage of Smokers:")
df['Smoking Numeric'] = df['Smoking'].map({'Yes': 1, 'No': 0})
print((df['Smoking Numeric'].mean()) * 100)


# 10) Scatter plot: BMI vs Triglyceride Level
plt.scatter(df['BMI'], df['Triglyceride Level'])
plt.xlabel("BMI")
plt.ylabel("Triglyceride Level")
plt.title("BMI vs Triglyceride Level")
plt.show()

# 11) Box plot: Sleep Hours
plt.boxplot(df['Sleep Hours'])
plt.ylabel("Hours")
plt.title("Sleep Hours Distribution")
plt.show()

# 12) Median sleep hours
print("\nMedian Sleep Hours:")
print(df['Sleep Hours'].median())
```
