```python
def factorial(n):
    if n < 0:
        raise ValueError("Factorial is not defined for negative numbers.")
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result
```

```python
print(factorial(7))
```

```python
def get_unique_elements(lst):
    unique = []
    for item in lst:
        if item not in unique:
            unique.append(item)
    return unique
```

```python
sample_list = [1,1,9,7,7,3,2]
unique_list = get_unique_elements(sample_list)
print(unique_list)
```

```python
def get_even_numbers(n):
    even_numbers = []
    num = 2
    while len(even_numbers) < n:
        even_numbers.append(num)
        num += 2
    print("First", n, "even numbers:", even_numbers)
    return even_numbers
n = 7
evens = get_even_numbers(n)
total = sum(evens)
product = 1
for number in evens:
    product *= number
print("Sum:", total)
print("Product:", product)
```

```python
def match_strings(str1, str2):
    if str1 == str2:
        return True
    else:
        return False

string1 = input("Enter first string: ")
string2 = input("Enter second string: ")

if match_strings(string1, string2):
    print("The strings match.")
else:
    print("The strings do not match.")
```
