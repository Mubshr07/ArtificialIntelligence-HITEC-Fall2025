```python
import heapq
#   BEST FIRST SEARCH 

def best_first_search(graph, h, start, goal):
    frontier = [(h[start], start)]  # (heuristic, node)
    visited = set()
    log = []

    while frontier:
        _, node = heapq.heappop(frontier)
        log.append(node)

        if node == goal:
            return log, node

        visited.add(node)
        for nxt, cost in graph[node]:
            if nxt not in visited:
                heapq.heappush(frontier, (h[nxt], nxt))

    return log, None


# ------------------------------
#   BEAM SEARCH
# ------------------------------
def beam_search(graph, h, start, goal, width=2):
    frontier = [start]
    visited = set()
    log = []

    while frontier:
        log.append(list(frontier))

        if goal in frontier:
            return log, goal

        candidates = []
        for node in frontier:
            visited.add(node)
            for nxt, cost in graph[node]:
                if nxt not in visited:
                    candidates.append((h[nxt], nxt))

        # keep only best 'width' nodes
        candidates.sort(key=lambda x: x[0])
        frontier = [n for _, n in candidates[:width]]

    return log, None


# ------------------------------
#   GRAPH (a) DEFINITION
# ------------------------------
graph_a = {
 'a': [('b',9), ('c',4), ('d',7)],
 'b': [('a',9), ('c',14), ('e',11)],
 'c': [('a',4), ('b',14), ('d',18), ('e',17), ('f',12)],
 'd': [('a',7), ('c',18), ('f',14)],
 'e': [('b',11), ('c',17), ('z',5)],
 'f': [('c',12), ('d',14), ('z',8)],
 'z': [('e',5), ('f',8)]
}

h_a = { 'a':21, 'b':14, 'c':18, 'd':18,
        'e':5, 'f':8, 'z':0 }

# Run searches
bfs_log_a, bfs_goal_a = best_first_search(graph_a, h_a, 'a', 'z')
beam_log_a, beam_goal_a = beam_search(graph_a, h_a, 'a', 'z', width=2)

print("Best First Search order:", bfs_log_a)
print("Beam Search levels:", beam_log_a)
print("--------------------------------\n")
```

```python
import heapq

# ------------------------------
#   BEST FIRST SEARCH 
# ------------------------------
def best_first_search(graph, h, start, goal):
    frontier = [(h[start], start)]
    visited = set()
    log = []

    while frontier:
        _, node = heapq.heappop(frontier)
        log.append(node)

        if node == goal:
            return log, node

        visited.add(node)
        for nxt, cost in graph[node]:
            if nxt not in visited:
                heapq.heappush(frontier, (h[nxt], nxt))

    return log, None


# ------------------------------
#   BEAM SEARCH
# ------------------------------
def beam_search(graph, h, start, goal, width=2):
    frontier = [start]
    visited = set()
    log = []

    while frontier:
        log.append(list(frontier))

        if goal in frontier:
            return log, goal

        candidates = []
        for node in frontier:
            visited.add(node)
            for nxt, cost in graph[node]:
                if nxt not in visited:
                    candidates.append((h[nxt], nxt))

        candidates.sort(key=lambda x: x[0])
        frontier = [n for _, n in candidates[:width]]

    return log, None


# ------------------------------
#   A* SEARCH
# ------------------------------
def a_star(graph, h, start, goal):
    frontier = [(h[start], 0, start)]   # (f, g, node)
    visited = {}
    log = []

    while frontier:
        f, g, node = heapq.heappop(frontier)
        log.append((node, g, f))

        if node == goal:
            return log, g

        if node in visited and visited[node] <= g:
            continue

        visited[node] = g

        for nxt, cost in graph[node]:
            new_g = g + cost
            new_f = new_g + h[nxt]
            heapq.heappush(frontier, (new_f, new_g, nxt))

    return log, None


# ------------------------------
#   GRAPH (b) DEFINITION
# ------------------------------
graph_b = {
 'S':[('A',4), ('B',10), ('C',11)],
 'A':[('B',8)],
 'B':[('D',15)],
 'C':[('D',8), ('E',20)],
 'D':[('H',16), ('F',1)],
 'H':[('I',2)],
 'I':[('J',5), ('K',13), ('D',20)],
 'F':[('G',19), ('I',5)],
 'E':[('G',19)],
 'J':[('G',7)],
 'K':[('G',16)],
 'G':[]
}

h_b = {
 'S':7, 'A':8, 'B':6, 'C':5, 'D':5,
 'E':3, 'F':3, 'G':0,
 'H':7, 'I':4, 'J':5, 'K':3
}

# Run searches
bfs_log_b, bfs_goal_b = best_first_search(graph_b, h_b, 'S', 'G')
beam_log_b, beam_goal_b = beam_search(graph_b, h_b, 'S', 'G', width=2)
astar_log_b, astar_cost_b = a_star(graph_b, h_b, 'S', 'G')

print("Best First Search order:", bfs_log_b)
print("Beam Search levels:", beam_log_b)
print("A* log:", astar_log_b)
print("A* final cost:", astar_cost_b)

```
