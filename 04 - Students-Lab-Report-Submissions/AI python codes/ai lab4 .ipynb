```python
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius ** 2

    def perimeter(self):
        return 2 * 3.14159 * self.radius
# Example usage 
if __name__ == "__main__":
    c = Circle(7) 
    print("Area:", c.area())
    print("Perimeter:", c.perimeter())
```

```python
class Power:
    def pow(self, x, n):
        if n == 0:
            return 1
        if n < 0:
            x = 1 / x
            n = -n
        result = 1
        while n:
            if n % 2 == 1:
                result *= x
            x *= x
            n //= 2
        return result
#example usage
if __name__ == "__main__":
    p = Power()

    print(p.pow(2, 3))    
    print(p.pow(5, 0))     
    print(p.pow(2, -2))    
    print(p.pow(10, 1))    
    print(p.pow(-2, 3))    
    print(p.pow(-2, 2))    
```

```python
class Item:
    def __init__(self, name, description, price):
        self.name = name
        self.description = description
        self.price = price

    def view_full_description(self):
        return f"Name: {self.name}\nDescription: {self.description}\nPrice: ${self.price:.2f}"

    def add_to_shopping_basket(self):
        return f"{self.name} added to shopping basket."

    def remove_from_shopping_basket(self):
        return f"{self.name} removed from shopping basket."


class MP3(Item):
    def __init__(self, name, description, price, artist, duration):
        super().__init__(name, description, price)
        self.artist = artist
        self.duration = duration

    def play(self):
        return f"Playing MP3: {self.name} by {self.artist}."

    def download(self):
        return f"Downloading MP3: {self.name}."


class DVD(Item):
    def __init__(self, name, description, price, certificate, duration, actors):
        super().__init__(name, description, price)
        self.certificate = certificate
        self.duration = duration
        self.actors = actors

    def view_trailer(self):
        return f"Viewing trailer for DVD: {self.name}."


class Book(Item):
    def __init__(self, name, description, price, author, number_of_pages, genre):
        super().__init__(name, description, price)
        self.author = author
        self.number_of_pages = number_of_pages
        self.genre = genre

    def preview_content(self):
        return f"Previewing content of book: {self.name} by {self.author}."
```

```python
if __name__ == "__main__":
    # example Test MP3
    mp3 = MP3("z41ns beats", "Relaxing lo-fi music", 3.99, "zain", "45 minutes")
    print(mp3.view_full_description())
    print(mp3.play())
    print(mp3.download())
    print(mp3.add_to_shopping_basket())
    print(mp3.remove_from_shopping_basket())

    print()  # Separator

    # Test DVD
    dvd = DVD("mr robot", "suspensful action movie", 14.99, "PG-13", "2 hours", ["Actor A", "Actor B"])
    print(dvd.view_full_description())
    print(dvd.view_trailer())

    print()  # Separator

    # Test Book
    book = Book("cyber security", "Learn cyber fundamentals", 24.99, "zain", 350, "Education")
    print(book.view_full_description())
    print(book.preview_content())
```
