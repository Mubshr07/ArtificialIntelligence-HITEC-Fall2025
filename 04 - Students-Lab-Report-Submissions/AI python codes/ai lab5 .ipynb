```python
import numpy as np

arr = np.array([[8, 3], [9, 1]])
print("Original flattened array:")
print(arr)
print("Maximum value of the above flattened array:", np.max(arr))
print("Minimum value of the above flattened array:", np.min(arr))

print("\n" + "-"*50 + "\n")
```

```python

array = np.array([1, 4, 9, 4])
print("Original array:", array)
float_array = array.astype(float)
print("Array converted to a float type:", float_array)
```

```python
matrix = np.arange(11,20).reshape(3, 3)
print("3x3 matrix with values ranging from 2 to 10:")
print(matrix)

print("\n" + "-"*50 + "\n")
```

```python
%config InlineBackend.print_figure_kwargs = {'bbox_inches':None}
import numpy as np
from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "all"

M1 = np.array([[5, 5],
               [1, 2]])  
M2 = np.array([[0, 1],
               [1, 0]])

resultant_matrix = np.dot(M1, M2)
transpose_matrix = resultant_matrix.T

print("Question 4:")
print("Matrix M1:\n", M1)
print("Matrix M2:\n", M2)
print("Resultant Matrix (M1 x M2):\n", resultant_matrix)
print("Transpose of Resultant Matrix:\n", transpose_matrix)
print("-" * 50)
```

```python

A = np.array([[6, 1, 1, 3],
              [4, -2, 5, 1],
              [2, 8, 7, 6],
              [3, 1, 9, 7]])

# Find inverse of matrix A
A_inverse = np.linalg.inv(A)

print("Question 5:")
print("Matrix A:\n", A)
print("Inverse of Matrix A:\n", A_inverse)

# (Optional) Verify that A * A_inverse = Identity matrix
identity_check = np.dot(A, A_inverse)
print("\nVerification (A x A_inverse):\n", identity_check)
```
