```python
# Import necessary libraries
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# ---------- STEP 1: Create Dataset ----------
# Features: Exercise hours per day, Water intake (liters per day)
# Target: Healthy (1) or Not healthy (0)
X_health = np.array([
    [1, 1.0],  # Low exercise, low water -> Not healthy
    [2, 2.0],  # Moderate exercise, moderate water -> Healthy
    [0.5, 0.5],# Very low exercise, low water -> Not healthy
    [3, 3.0],  # High exercise, high water -> Healthy
    [2, 1.5]   # Moderate exercise, moderate water -> Healthy
])

y_health = np.array([0, 1, 0, 1, 1])

# ---------- STEP 2: Build ANN Model ----------
model_health = Sequential()
model_health.add(Dense(5, input_dim=2, activation='relu'))  # Hidden layer with 5 neurons
model_health.add(Dense(1, activation='sigmoid'))            # Output layer (1 neuron for binary classification)

# ---------- STEP 3: Compile Model ----------
model_health.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])

# ---------- STEP 4: Train Model ----------
model_health.fit(X_health, y_health, epochs=100, verbose=0)

# ---------- STEP 5: Make Prediction ----------
# Predict if a person exercising 2 hours and drinking 2 liters of water is healthy
prediction = model_health.predict(np.array([[2, 2]]))
print("Prediction (Health):", int(prediction[0][0] > 0.5))  # Convert probability to 0 or 1
```

```python
# ---------- STEP 1: Create Dataset ----------
# Features: Assignments submitted, Attendance percentage
# Target: Pass (1) or Fail (0)
X_student = np.array([
    [5, 90],   # High assignments, high attendance -> Pass
    [3, 60],   # Moderate assignments, moderate attendance -> Pass
    [1, 40],   # Low assignments, low attendance -> Fail
    [4, 70],   # Moderate-high assignments & attendance -> Pass
    [2, 50]    # Low assignments, low attendance -> Fail
])

y_student = np.array([1, 1, 0, 1, 0])

# ---------- STEP 2: Build ANN Model ----------
model_student = Sequential()
model_student.add(Dense(5, input_dim=2, activation='relu'))  # Hidden layer
model_student.add(Dense(1, activation='sigmoid'))            # Output layer

# ---------- STEP 3: Compile Model ----------
model_student.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])

# ---------- STEP 4: Train Model ----------
model_student.fit(X_student, y_student, epochs=100, verbose=0)

# ---------- STEP 5: Make Prediction ----------
# Predict if a student submitting 3 assignments and 65% attendance will pass
prediction_student = model_student.predict(np.array([[3, 65]]))
print("Prediction (Student Pass/Fail):", int(prediction_student[0][0] > 0.5))
```

```python
# ---------- STEP 1: Create Dataset ----------
# Features: Monthly income (in $1000s), Credit score
# Target: Eligible for loan (1) or Not (0)
X_loan = np.array([
    [3, 600],  # Low income, low credit score -> Not eligible
    [5, 650],  # Moderate income, moderate credit score -> Eligible
    [2, 550],  # Low income, low credit score -> Not eligible
    [8, 700],  # High income, good credit -> Eligible
    [4, 620]   # Moderate income, moderate credit -> Eligible
])

y_loan = np.array([0, 1, 0, 1, 1])

# ---------- STEP 2: Build ANN Model ----------
model_loan = Sequential()
model_loan.add(Dense(5, input_dim=2, activation='relu'))  # Hidden layer
model_loan.add(Dense(1, activation='sigmoid'))            # Output layer

# ---------- STEP 3: Compile Model ----------
model_loan.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])

# ---------- STEP 4: Train Model ----------
model_loan.fit(X_loan, y_loan, epochs=100, verbose=0)

# ---------- STEP 5: Make Prediction ----------
# Predict if a person with $4k income and 640 credit score is eligible
prediction_loan = model_loan.predict(np.array([[4, 640]]))
print("Prediction (Loan Eligibility):", int(prediction_loan[0][0] > 0.5))
```
